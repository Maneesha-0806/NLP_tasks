```python
import re
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

data = pd.read_csv(r"C:\Users\manee\Documents\NLP_skill\home_tasks\sentiment.csv", nrows = 5000)
print("----------Sample Data---------")
print(data.head())

print("\n-----------------")
print("Number of duplicates:", data.duplicated(subset=['review']).sum())
print("Missing values per column:\n", data.isnull().sum())
data = data.drop_duplicates(subset = ['review']).reset_index(drop = True)
data = data.dropna()

stop_words = set(stopwords.words('english'))
lemma = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r'<.*?>', ' ' ,text)#remove HTML tags
    text = re.sub(r'http\S+|www\.\S+', ' ', text)#remove URLs
    text = re.sub(r'[^a-z\s]', ' ', text)#remove non-alphabetic characters
    text = re.sub(r'\s+', ' ', text).strip()#Normalize whitespace

    tokens = word_tokenize(text)
    cleaned_tokens = [
        lemma.lemmatize(word) 
        for word in tokens 
        if word not in stop_words and len(word) > 1
    ]
    return " ".join(cleaned_tokens)

data['cleaned_review'] = data['review'].apply(preprocess)

print("\n----------Original----------")
print(data['review'].iloc[0][:300])
print("----------Cleaned----------")
print(data['cleaned_review'].iloc[0][:300])

data.to_csv('Cleaned_IMDB_Sentiment_analysis_data.csv', index = False)
print("\n----------Exported as CSV----------")
```
